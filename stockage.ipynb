```python
df_companies = spark.read.csv(
    "/Volumes/workspace/default/my_volume/companies_clean_final.csv",
    header=True,
    inferSchema=True
)

# Renommer les colonnes problématiques
df_companies = df_companies \
    .withColumnRenamed("Annual_Revenue (M₺)", "Annual_Revenue_M") \
    .withColumnRenamed("Marketing_Spend (K₺)", "Marketing_Spend_K") \
    .withColumnRenamed("Conversion_Rate (%)", "Conversion_Rate_Pct")

df_companies.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.companies")
```

```python
spark.table("workspace.default.companies").show(10)
```
